## Problems We Fix In This Notebook
1. Drop `customerID` — useless for ML
2. Fix `TotalCharges` — convert from text to number
3. Drop 11 rows where `TotalCharges` is empty
4. Convert `Churn` from Yes/No to 1/0
5. Save clean data to `data/processed/`

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv')

print(" Data loaded")
print(f"Shape: {df.shape}")
```

```python
df.drop(columns=['customerID'], inplace=True)
print(f"Shape after dropping customerID: {df.shape}")
```

```python
df['TotalCharges']=pd.to_numeric(df['TotalCharges'], errors='coerce')
print(f"TotalCharges dtype: {df['TotalCharges'].dtype}")
print(f"Missing values in TotalCharges: {df['TotalCharges'].isnull().sum()}")
```

```python
df=df.dropna(subset=['TotalCharges'])
print(f"Shape after dropping missing rows: {df.shape}")
```

```python
print(f"Duplicate rows: {df.duplicated().sum()}")
```

```python
df = df.drop_duplicates()

print(f"Shape after dropping duplicates: {df.shape}")
```

```python
print(f"Churn unique values: {df['Churn'].unique()}")
print(f"\nChurn value counts:\n{df['Churn'].value_counts()}")
```

```python
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})

print(f"Churn after mapping:\n{df['Churn'].value_counts()}")
```

```python
df.to_csv('../data/processed/telco_cleaned.csv', index=False)

print(" Cleaned data saved to data/processed/telco_cleaned.csv")
print(f"Final shape: {df.shape}")
```

## Summary — What We Cleaned
- Dropped `customerID` — useless for ML
- Converted `TotalCharges` from text to number
- Dropped 11 rows with empty `TotalCharges` (tenure = 0 customers)
- Dropped 22 duplicate rows
- Converted `Churn` from Yes/No to 1/0

## Important Note — Class Imbalance
- 73% customers stayed (0)
- 27% customers churned (1)
- Dataset is imbalanced — we will handle this during model building

## Final Clean Data
- 7010 rows, 20 columns
- Saved to `data/processed/telco_cleaned.csv`
