## What We Do In This Notebook
1. Load cleaned data
2. Drop TotalCharges (highly correlated with tenure)
3. Encode binary categorical columns (Label Encoding)
4. Encode multi-category columns (One Hot Encoding)
5. Scale numerical columns
6. Save final prepared data

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/processed/telco_cleaned.csv')

print("Data loaded")
print(f"Shape: {df.shape}")
```

```python
df = df.drop(columns=['TotalCharges'])

print(f"Shape after dropping TotalCharges: {df.shape}")
```

```python
df.dtypes
```

## Step 1 — Label Encoding
Columns with only 2 categories → convert directly to 0 and 1.

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

binary_cols = ['gender', 'Partner', 'Dependents', 
               'PhoneService', 'PaperlessBilling']

for col in binary_cols:
    df[col] = le.fit_transform(df[col])
    
print("✅ Label encoding complete")
print(df[binary_cols].head())
```

## Step 2 — One Hot Encoding
Columns with 3+ categories → create separate column for each category.
This avoids giving false ranking to categories.

```python
ohe_cols = ['MultipleLines', 'InternetService', 'OnlineSecurity',
            'OnlineBackup', 'DeviceProtection', 'TechSupport',
            'StreamingTV', 'StreamingMovies', 'Contract', 'PaymentMethod']

df = pd.get_dummies(df, columns=ohe_cols, drop_first=True)

print("✅ One Hot Encoding complete")
print(f"Shape after encoding: {df.shape}")
```

## Step 3 — Scaling Numerical Columns
Bring all numerical columns to the same range (0-1).
This ensures no single feature dominates the model
just because of its scale.

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

numerical_cols = ['tenure', 'MonthlyCharges']

df[numerical_cols] = scaler.fit_transform(df[numerical_cols])

print(" Scaling complete")
print(df[numerical_cols].describe())
```

## Step 4 — Save Prepared Data
Save the final encoded and scaled data to `data/processed/`
This is the file our ML model will train on directly.

```python
df.to_csv('../data/processed/telco_prepared.csv', index=False)

print("✅ Prepared data saved to data/processed/telco_prepared.csv")
print(f"Final shape: {df.shape}")
print(f"\nAll columns:\n{list(df.columns)}")
```

## Summary — Feature Engineering Complete

### What We Did
- Dropped `TotalCharges` — highly correlated with tenure (0.83)
- Label Encoded 5 binary columns → 0 and 1
- One Hot Encoded 10 multi-category columns → 11 new binary columns
- Scaled `tenure` and `MonthlyCharges` → range 0 to 1

### Final Data
- 7010 rows, 30 columns
- All columns are numbers — ready for ML model
- Saved to `data/processed/telco_prepared.csv`
